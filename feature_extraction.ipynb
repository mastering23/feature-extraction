# Feature Extraction 🔩
- Most classic machine learning algorithms can't take in raw text.
- Instead, we need to perform a feature "extraction" from the raw text to pass numerical features to the machine learning algorithm.
    - Counter Vectorization | Term-Frequency | Document Frequency.

````python
# Count Vectorization
messages = ["Hey, lets go to the game today!",
           "Call your sister.",
           "Want to go walk your dogs?"]
````

````python
from sklearn.feature_extraction.text import CountVectorizer
vect = CountVectorizer()
````

````python
vect = CountVectorizer()
X = vect.fit_transform(messages)
````

````python
print(X.toarray())
````

````python
print(vect.get_feature_names_out())
````

- An alternative to CountVectorizer is TfidfVectorizer. It also creates a document term matrix from our messages.
- However, instead of filling the DTM with token counts, it calculates term frequency-inverse document frequency value for each word(TF-IDF).
    ```Term frequency tf(t,d): is the raw count of a term in a document, i.e. the number of times that term t occurs in the document d.```

#### Ordinal Categorical Data:Context

- How satisfied are you with the service?
    - Very dissatisfied
    - Dissatisfied
    - Neutral
    - Satisfied
    - Very satisfied
- How would you rate the quality of the food?
    - Bad
    - Good
    - Very Good
    - Excellent

````python
import pandas as pd
import numpy as np

service = ["Very dissatisfied", "Dissatisfied", "Neutral", "Satisfied", "Very satisfied"]
quality = ["Bad", "Good", "Very Good", "Excellent"]

survey = {
    "services": ["Very dissatisfied", "Dissatisfied", "Neutral", "Satisfied", "Very satisfied", "Very dissatisfied"],
    "food": ["Bad", "Good", "Very Good", "Excellent", "Good", "Bad"]  
}

type_client = [0, 0, 1, 1, 0, 1]

pd.DataFrame(survey)
````

### Ordinal Econder

````python
from sklearn.preprocessing import OrdinalEncoder
data_ord = pd.DataFrame(survey)
codefider = OrdinalEncoder(categories =[service,quality])
data_ord = pd.DataFrame(codefider.fit_transform(data_ord),columns=["service","food"])
print(data_ord)
print("\nCategories\n")
print(codefider.categories_)
````

### Codefider OneHotEncoder

````python
from sklearn.preprocessing import OneHotEncoder
datos_one = pd.DataFrame(survey)
codefider = OneHotEncoder()
print(codefider.fit_transform(datos_one))
````

````python
print(codefider.fit_transform(datos_one).toarray())
````

````python
data_one = pd.DataFrame(codefider.fit_transform(datos_one).toarray(),columns = np.concatenate(codefider.categories_))
data_one
````
